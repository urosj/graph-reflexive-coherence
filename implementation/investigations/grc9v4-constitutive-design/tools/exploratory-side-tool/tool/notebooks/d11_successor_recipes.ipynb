# GRCv4 D11 successor forensic recipes

This notebook exposes accepted D11-C and D11-G9 authority through the pure forensic API. It writes only derived evidence traces and does not create scientific authority.

```python
from grcv4_explorer.forensic import contract_provenance, debt_lifecycle, reconstruction_path, write_trace
from grcv4_explorer.paths import SIDE_TOOL_ROOT, TOOL_ROOT, repository_root
from grcv4_explorer.successor import load_successor_forensic_context
repo_root = globals().get('repo_root', repository_root())
side_tool_root = globals().get('side_tool_root', SIDE_TOOL_ROOT)
output_dir = globals().get('output_dir', TOOL_ROOT / 'generated/iteration11-notebook')
output_dir.mkdir(parents=True, exist_ok=True)
context = load_successor_forensic_context(repo_root, side_tool_root)
```

```python
write_trace(output_dir / 'd11-c-claim.json', reconstruction_path(context, 'D11-C-CL-O-001'))
write_trace(output_dir / 'd11-c-debt.json', debt_lifecycle(context, 'D11-C-DEBT-BASELINE-TRANSPORT-AUTHORITY'))
write_trace(output_dir / 'd11-c-contract.json', contract_provenance(context, 'D11-C-EC-C-J0-CURRENT'))
```
